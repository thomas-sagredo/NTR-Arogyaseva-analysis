```python
from src.utils.io import load_data, set_pandas_display_options

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Establecer opciones de visualización para pandas
set_pandas_display_options()

# Importar el dataset
df = load_data(stage="clean")
```

```python
df.set_index("", inplace=True)
```

```python
df.info()
```

```python
inconsistent_surgery_mask = df["inconsistent_surgery_to_category_name"]
print("Número de filas con relación cirugía-categoría inconsistente:", inconsistent_surgery_mask.sum())
print("\nCirugías con múltiples categorías:\n")
print(df.loc[inconsistent_surgery_mask, "surgery"].value_counts())
```

```python
print("Cirugías y sus categorías asociadas:\n")
print(df.loc[inconsistent_surgery_mask, ["surgery", "category_name"]].value_counts().sort_index(ascending=False))
```
